##**Project** **Objective**

###The objective of this analysis is to identify the major factors contributing to employee attrition and provide actionable recommendations to improve employee retention.

#**Key** **Findings**

1. Overall attrition rate is 16.12%.

2. Employees working overtime show significantly higher attrition (30.53%) than employees not working overtime (10.44%).

3. Employees earning below 5k and working overtime exhibit attrition rates above 42%.

4. Employees aged 18-25 have the highest attrition rate (35.77%).

5. Sales Representatives working overtime show the highest attrition rate (66.67%).

6. Single employees are more likely to leave than married employees.

7. Poor work-life balance is associated with higher employee turnover.

#**Business** **Impact** **Analysis**

##Finding 1

High Overtime Attrition

## Business Impact

Frequent employee turnover increases recruitment, onboarding, and training costs while reducing team productivity.

## Finding 2

Low Salary + Overtime

## Business Impact

Employees may perceive workload and compensation as imbalanced, increasing the likelihood of resignation.

## Finding 3
Young Employees Leaving

## Business Impact

Loss of early-career talent can create long-term workforce shortages and increase hiring requirements.

# **Recommendations**

| Issue                          | Recommendation                                            |
| ------------------------------ | --------------------------------------------------------- |
| High Overtime                  | Monitor overtime hours and redistribute workload          |
| Low Salary Attrition           | Review compensation structure for lower salary bands      |
| Young Employee Attrition       | Create mentorship and career growth programs              |
| Sales Representative Attrition | Investigate workload, targets, and incentive plans        |
| Poor Work-Life Balance         | Introduce flexible work policies and wellness initiatives |

# **Executive Summary**

Employee attrition analysis identified overtime, low salary, age group, and job role as major drivers of employee turnover. Employees working overtime were nearly three times more likely to leave the organization. The highest-risk segment consisted of Sales Representatives working overtime, with attrition exceeding 66%. Strategic interventions focused on workload management, compensation review, career development, and employee well-being are recommended to improve retention.
